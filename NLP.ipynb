```python
import spacy
from spacy import displacy
```

```python
text="Makemake is a dwarf planet orbiting the Sun beyond Neptune. It has a diameter 60% that of Pluto, and is the fourth-largest trans-Neptunian object and the largest member of the Solar System's classical Kuiper belt, a disk of icy bodies beyond Neptune's orbit. Its discovery on March 31, 2005, by American astronomers Mike Brown, Chad Trujillo and David Rabinowitz at Palomar Observatory contributed to the reclassification of Pluto as a dwarf planet in 2006. Makemake's surface, like Pluto's, is largely covered by frozen methane and stained reddish-brown by tholins. It has one known satellite, unnamed, whose orbit suggests that Makemake's rotation has a high axial tilt. Makemake shows evidence of geochemical activity and cryovolcanism, which has led scientists to suspect that it might harbor a subsurface ocean of liquid water. No high-resolution images of its surface exist because it has not been visited up close by a space probe."
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
for token in text.split():
    print(token)
```

```python
doc=nlp(text)

for token in doc[:100]:
    print(token)
```

```python
print(len(text))
```

```python
print(len(doc))
```

```python
sentence1=doc.sents

print(sentence1)
print(list(sentence1)[0])
```

```python
text1="Tejas is my friend."
doc2=nlp(text1)

print(doc2)
```

```python
for token in doc2:
    print(token.text,token.pos_,token.dep_)
```
